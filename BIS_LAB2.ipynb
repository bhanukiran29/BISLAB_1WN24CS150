```python
import random


# Profit function
def profit(price):
    demand = 100 - 2 * price
    return (price - 20) * demand


# Initial prices
p = [random.uniform(20, 50) for _ in range(5)]

# Initial velocities
v = [0] * 5

# Find the initial best price
best = max(p, key=profit)


# Optimization
for _ in range(20):

    for i in range(5):

        # Update velocity
        v[i] += 1.5 * random.random() * (best - p[i])

        # Update position (price)
        p[i] += v[i]

        # Update global best
        if profit(p[i]) > profit(best):
            best = p[i]


# Output
print("Best price:", round(best, 2))
print("Maximum profit:", round(profit(best), 2))
print("")
print("Executed by Madisetty Bhanu Kiran")
```
