```python

import random
import string

# Genetic Algorithm Parameters
POPULATION_SIZE = 100
MUTATION_RATE = 0.1
GENERATIONS = 1000
TARGET = "Hello GA!"

# Valid characters
GENES = string.ascii_letters + string.digits + " !.,"

# Individual class
class Individual:

    def __init__(self, chromosome):
        self.chromosome = chromosome
        self.fitness = self.calculate_fitness()

    # Generate a random chromosome
    @staticmethod
    def create_chromosome():
        return ''.join(
            random.choice(GENES)
            for _ in range(len(TARGET))
        )

    # Calculate fitness
    def calculate_fitness(self):
        return sum(
            1 for a, b in zip(self.chromosome, TARGET)
            if a != b
        )

    # Crossover operation
    def crossover(self, parent2):
        child = []

        for gene1, gene2 in zip(
            self.chromosome, parent2.chromosome
        ):
            if random.random() < 0.5:
                child.append(gene1)
            else:
                child.append(gene2)

        return Individual(''.join(child))

    # Mutation operation
    def mutate(self):
        chromosome = list(self.chromosome)

        for i in range(len(chromosome)):
            if random.random() < MUTATION_RATE:
                chromosome[i] = random.choice(GENES)

        return Individual(''.join(chromosome))


# Main Genetic Algorithm
def genetic_algorithm():

    # Initial population
    population = [
        Individual(Individual.create_chromosome())
        for _ in range(POPULATION_SIZE)
    ]

    generation = 0

    while generation < GENERATIONS:

        # Sort population by fitness
        population.sort(key=lambda x: x.fitness)

        best = population[0]

        print(
            f"Generation: {generation} | "
            f"String: {best.chromosome} | "
            f"Fitness: {best.fitness}"
        )

        # Check termination condition
        if best.fitness == 0:
            print("\nTarget string successfully generated!")
            break

        # Elitism: retain best 10 individuals
        new_population = population[:10]

        # Generate remaining offspring
        while len(new_population) < POPULATION_SIZE:

            # Parent selection
            parent1 = random.choice(population[:50])
            parent2 = random.choice(population[:50])

            # Crossover
            child = parent1.crossover(parent2)

            # Mutation
            child = child.mutate()

            new_population.append(child)

        population = new_population
        generation += 1

    print("\nFinal Result:")
    print("Target:", TARGET)
    print("Generated:", population[0].chromosome)
    print("Fitness:", population[0].fitness)


# Driver code
if __name__ == "__main__":
    genetic_algorithm()
```
