```python
import random
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score


# Load dataset
data = load_iris()
X = data.data
y = data.target

# Parameters
population_size = 10
genes = X.shape[1]
mutation_rate = 0.1
crossover_rate = 0.8
generations = 20


# Fitness function
def fitness(chromosome):
    selected = [i for i in range(genes) if chromosome[i] == 1]

    # Avoid empty feature set
    if len(selected) == 0:
        return 0

    X_selected = X[:, selected]

    model = DecisionTreeClassifier(random_state=42)

    score = cross_val_score(
        model, X_selected, y, cv=5
    ).mean()

    return score


# Generate random population
def random_population(size):
    return [
        [random.randint(0, 1) for _ in range(genes)]
        for _ in range(size)
    ]


# Selection
def select_best(population, scores):
    ranked = sorted(
        zip(population, scores),
        key=lambda x: x[1],
        reverse=True
    )

    return [chromosome for chromosome, score in ranked[:population_size // 2]]


# Crossover
def crossover(parent1, parent2):
    point = random.randint(1, genes - 1)

    child = parent1[:point] + parent2[point:]

    return child


# Mutation
def mutate(chromosome):
    for i in range(genes):
        if random.random() < mutation_rate:
            chromosome[i] = 1 - chromosome[i]


# Initialize population
population = random_population(population_size)

best_solution = None
best_fitness = 0


# Main GEA
for generation in range(generations):

    # Gene Expression + Fitness
    scores = []

    for chromosome in population:
        score = fitness(chromosome)
        scores.append(score)

        if score > best_fitness:
            best_fitness = score
            best_solution = chromosome.copy()

    # Selection
    selected = select_best(population, scores)

    # Crossover
    new_population = []

    while len(new_population) < population_size:

        parent1 = random.choice(selected)
        parent2 = random.choice(selected)

        if random.random() < crossover_rate:
            child = crossover(parent1, parent2)
        else:
            child = parent1.copy()

        new_population.append(child)

    # Mutation
    for chromosome in new_population:
        mutate(chromosome)

    # New generation
    population = new_population


# Final result
selected_features = [
    data.feature_names[i]
    for i in range(genes)
    if best_solution[i] == 1
]

print("Best Feature Selection:", best_solution)
print("Selected Features:", selected_features)
print("Best Fitness (Accuracy):", best_fitness)
print("")
print("Executed by Madisetty Bhanu Kiran")
```
